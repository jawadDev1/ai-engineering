### Nakli Chain

```python
import random

class NakliLLM:
    def __init__(self):
        print("LLM created")

    def predict(self, prompt):
        res_list = [
            "Islamabad is the capital of Pakistan",
            "Berlin is the capital of Germany",
            "Paris is the capital of France",
        ]


        return {"response": random.choice(res_list)}
```

```python
llm = NakliLLM()

llm.predict("What is the capital of pakistan")
```

```python
class NakliPromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):
        return self.template.format(**input_dict)

template = NakliPromptTemplate("Write a {length} poem about {topic}", ["topic", "length"])

template.format({"topic": "anime", "length": "short"})
```

```python

prompt = template.format({"topic": "anime", "length": "short"})
llm.predict(prompt)
```

```python
class NakliLLMChain:
    def __init__(self, prompt, llm):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        prompt = self.prompt.format(input_dict)
        result =  self.llm.predict(prompt)
        print(result)
        return result['response']
```

```python
chain = NakliLLMChain(template, llm)

chain.run({"topic": "anime", "length": "short"})
```
