```python
from abc import ABC, abstractmethod 
```

```python
class Runnable(ABC):

    @abstractmethod
    def invoke(input_method):
        pass
```

```python

import random

class NakliLLM(Runnable):
    def __init__(self):
        print("LLM created")

    def invoke(self, prompt):
        res_list = [
            "Islamabad is the capital of Pakistan",
            "Berlin is the capital of Germany",
            "Paris is the capital of France",
        ]


        return {"response": random.choice(res_list)}

    def predict(self, prompt):
        res_list = [
            "Islamabad is the capital of Pakistan",
            "Berlin is the capital of Germany",
            "Paris is the capital of France",
        ]


        return {"response": random.choice(res_list)}
```

```python
llm = NakliLLM()
```

```python

class NakliPromptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, input_dict):
        return self.template.format(**input_dict)

    def format(self, input_dict):
        return self.template.format(**input_dict)

template = NakliPromptTemplate("Write a {length} poem about {topic}", ["topic", "length"])

template.format({"topic": "anime", "length": "short"})
```

```python
class RunnalbeConnector(Runnable):

    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self, input_data):
        
        for runnable in self.runnable_list:
            print("before ", input_data)
            input_data = runnable.invoke(input_data)
        return input_data
```

```python
chain = RunnalbeConnector([template, llm])

result = chain.invoke({"topic": "anime", "length": "short"})
print(result)
```
